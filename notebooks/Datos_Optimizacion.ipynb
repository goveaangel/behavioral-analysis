```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Diccionario para convertir números a nombres de meses
month_names = {
    1: 'January', 2: 'February', 3: 'March', 4: 'April',
    5: 'May', 6: 'June', 7: 'July', 8: 'August',
    9: 'September', 10: 'October', 11: 'November', 12: 'December'
}

# Convertimos los números a nombres de mes
data['Month_Name'] = data['Month'].map(month_names)

# Calculamos el promedio de engagement por mes
engagement_month = data.groupby('Month_Name')['Engagement'].mean().reset_index()

# Aseguramos el orden cronológico
engagement_month['Month_Name'] = pd.Categorical(
    engagement_month['Month_Name'],
    categories=list(month_names.values()),
    ordered=True
)

# Configuración del gráfico
sns.set(style="whitegrid", palette="coolwarm")
plt.figure(figsize=(10, 6))

sns.lineplot(data=engagement_month, x='Month_Name', y='Engagement', marker='o')

plt.title('Engagement promedio a lo largo de los meses', fontsize=14)
plt.xlabel('Mes', fontsize=12)
plt.ylabel('Engagement promedio', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
